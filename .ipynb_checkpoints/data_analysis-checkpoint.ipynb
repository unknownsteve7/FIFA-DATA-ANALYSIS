DATA Collection

```python
import pandas as pd

editions = pd.read_csv('wc_all_editions.csv')
matches = pd.read_csv('wc_all_matches.csv')
scorers = pd.read_csv('wc_top_scorers.csv')
fixtures = pd.read_csv('wc_2026_fixtures.csv')
teams = pd.read_csv('wc_2026_teams.csv')
```

```python
from sqlalchemy import create_engine
import pandas as pd

engine = create_engine('mysql+pymysql://root:1105@localhost/fifa_football')



editions.to_sql('editions', engine, if_exists='replace', index=False)
matches.to_sql('matches', engine, if_exists='replace', index=False)
scorers.to_sql('scorers', engine, if_exists='replace', index=False)
fixtures.to_sql('fixtures', engine, if_exists='replace', index=False)
teams.to_sql('teams', engine, if_exists='replace', index=False)

print("Done — all tables loaded!")
```
